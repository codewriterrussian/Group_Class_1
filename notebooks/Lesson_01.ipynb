這一課先熟悉 Python 最基本的兩個操作：

- `print()`：把內容輸出到畫面上
- `input()`：讓使用者輸入內容

## 1. `print()`：輸出內容

`print()` 可以把文字、數字或其他資料顯示在畫面上。

先從最經典的 **Hello World** 開始。

````python
print("hello world")
````

## 2. `input()`：接收使用者輸入

`input()` 會暫停程式，等待使用者輸入內容。

使用者按下 Enter 後，`input()` 會把輸入內容回傳給程式。

> 在 Jupyter Notebook 中執行 `input()` 時，輸入框會直接出現在 cell 的執行區域。

````python
input()
````

### 輸入後立刻輸出

可以把 `input()` 放在 `print()` 裡：

1. `input()` 先取得使用者輸入
2. `print()` 再把輸入內容印出來

````python
print(input())
````

## 3. `input()` 和 `print()` 分開使用

下面的程式會先等待輸入，但因為沒有把輸入結果交給 `print()`，所以最後只會印出一個空行。

````python
input()
print()
````

## 4. 一次接收兩個輸入

可以在同一個 `print()` 裡放兩個 `input()`：

```python
print(input(), input())
```

Python 會依序等待兩次輸入，再把兩個結果印出來。

````python
print(input(), input())
````

## 5. 固定文字 + 使用者輸入

`print()` 可以同時輸出固定文字與 `input()` 的結果。

````python
print("hi,", input())
````

````python
print("hello,", input())
````

## 6. 微波爐比喻

可以把程式想成一台「微波爐」：

- `input()`：把資料放進程式
- 程式處理：程式內部做事情
- `print()`：把結果拿出來、顯示給使用者

## 7. `input("提示文字")`

`input()` 裡也可以放提示文字，例如：

```python
input("Input: ")
```

執行時，畫面會先顯示 `Input: `，再等待使用者輸入。

要特別注意：

- `"Input: "` 是提示文字
- 使用者實際輸入的內容才是 `input()` 的回傳值
- `input()` 的回傳值型別預設是字串（`str`）
- 如果沒有把回傳值存起來，之後就無法再次使用它

````python
input("Input: ")
````

### 把輸入存進變數

通常會把 `input()` 的結果存進變數，之後再使用。

````python
user_input = input("Input: ")
print(user_input)
````

## 8. 延伸：搭配 `if`

原本的範例使用了：

```python
if i == 0:
```

因此執行前必須先讓 `i` 有值。下面補上 `i = 0`，讓範例可以直接執行。

````python
i = 0

if i == 0:
    print("hi,", input())
    print("hello,", input())
````
