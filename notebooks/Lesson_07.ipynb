```python
x=int(input())

if x >= 80:
    if x >= 90:         print('A+')
    elif x >= 85:       print('A')
    else:               print('A-')
elif x >= 70:
    if x >= 77:         print('B+')
    elif x >= 73:       print('B')
    else:               print('B-')
elif x >= 60:
    if x >= 67:         print('C+')
    elif x >= 63:       print('C')
    else:               print('C-')
elif x > 0:             print('F')
else:                   print('X')
```
