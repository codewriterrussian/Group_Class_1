```python
# 運算 -> 算錢
# + - * / -> 加減乘除

print(2/3) 

# 額外 -> ** // %

print( 2**2 == 2*2 ) # 2^2 -> 4 == 4
print( 3**3 == 3*3*3) # 3^3 = 27 | 9

# 20 除以 3 🟰 6 餘 2
print(20//3)
print(20%3)
```
