```python
# sep
print(1,2,3,4,5)
print(1,2,3,4,5, sep=',')

# '' 保留 -> 給人類看的字 / 沒寫‘’ 那就是變數

# 練習: 蛋餅/奶茶/水果
print("蛋餅","奶茶","水果",sep='/')

# 練習: sep=' -> ' (3 個地名)
print("台北","台中","嘉義" ,sep=' -> ')
```

```python
print(1,2,3,4,5, sep=' ')
print(1,2,3,4,5) # 預設 -> sep=' '
```

```python
print(123, end=' ')
print(456)
# 預設 -> end='\n'

print("蛋餅","奶茶","水果",sep='/')
#改成用 end -> 3 個print
print("蛋餅",end='/')
print("奶茶",end='/')
print("水果",end='')
```

```python
name, age, city = input().split()

first_line = "我的名字是 " + name
second_line = "年齡是 " + age
third_line = "住在 " + city


# 結果要印出3行
# print("我的名字是",name)
# print("年齡是",age)
# print("住在",city)
print(first_line, second_line, third_line, sep='\n')
```
