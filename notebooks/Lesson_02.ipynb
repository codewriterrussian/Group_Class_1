## 1. 變數（Variable）

`=` 是**指定值**，不是數學中的等號。再次指定同一個變數時，新值會取代舊值。

````python
name = "John"
name = "Mary"  # 更新 name 的值
print(name)      # Mary
````

字串需要引號，例如 `"Mary"`；沒有引號的 `name` 是變數名稱。

````python
a = "hi"   # 字串（str）
b = 10     # 整數（int）

print(a)
print(b)
````

````python
a = 10
b = 'a'

a = b

print(a)
print(b)
````

## 2. `print()`：輸出資料

用逗號分隔多個輸出項目，預設會在項目之間放一個空格。

````python
print(123, "早安")
print(1, 2, 3, 4, 5)
````

### 2.1 `sep`：設定項目之間的分隔符號

- `sep=" "`：預設以空格分隔。
- `sep=","`：改用逗號分隔。
- `sep="/"`：改用斜線分隔。

````python
print(1, 2, 3, 4, 5)             # 預設使用空格
print(1, 2, 3, 4, 5, sep=" ")    # 與上一行相同
print(1, 2, 3, 4, 5, sep=",")

print("蛋餅", "奶茶", "水果", sep="/")
print("台北", "台中", "嘉義", sep=" -> ")
````

### 2.2 `end`：設定一次 `print()` 結尾的字元

- 預設為 `end="\n"`（換行）。
- `end=" "` 會在行尾放空格，不立即換行。
- `end=""` 代表不增加任何結尾字元。

````python
print(123, end=" ")
print(456)  # 與上一行接在同一行

print("蛋餅", "奶茶", "水果", sep="/")

# 使用三個 print() 輸出相同內容
print("蛋餅", end="/")
print("奶茶", end="/")
print("水果")
````

## 3. `input()`：讀取使用者輸入

`input()` 會等待使用者輸入文字；回傳值的型別是**字串** `str`。即使輸入 `13`，一開始也會當作字串。

````python
data = input("請輸入一段文字：")
print("你輸入了：", data)
````

### 3.1 `split()`：拆開一行資料

例如輸入：

```text
小明 13 台北
```

`input().split()` 會根據空白把輸入分成三段，再依序指定給 `name`、`age`、`city`。這裡的 `age` **仍是字串**。

````python
# 請輸入：小明 13 台北
name, age, city = input("請輸入姓名 年齡 城市：").split()
print(name, age, city)
````

### 3.2 組合文字並輸出三行

使用 `+` 連接字串，並用 `sep="\n"` 讓三段內容分成三行。

````python
# 請輸入：小明 13 台北
name, age, city = input("請輸入姓名 年齡 城市：").split()

first_line = "我的名字是 " + name
second_line = "年齡是 " + age
third_line = "住在 " + city

print(first_line, second_line, third_line, sep="\n")
````

預期輸出（輸入為 `小明 13 台北`）：

```text
我的名字是 小明
年齡是 13
住在 台北
```

## 4. 型別（Type）與型別轉換

- `10` 是整數（`int`），可直接做加法。
- `"10"` 是字串（`str`），與另一個字串相加會**串接文字**。
- `int("10")`：將數字字串轉為整數。
- `str(10)`：將整數轉為字串。

````python
num = 10
num_2 = "10"

print(num + 10)          # 20：整數加法
print(int(num_2) + 10)   # 20：先轉整數再相加
print(str(num) + "10")   # 1010：字串串接
````
