## 2. 比較運算子

比較運算的結果是 `True` 或 `False`。

| 運算子 | 意義 | 範例 |
| :---: | --- | --- |
| `==` | 相等 | `3 == 3` |
| `!=` | 不相等 | `3 != 3` |
| `<` | 小於 | `3 < 3` |
| `>` | 大於 | `3 > 3` |
| `<=` | 小於等於 | `3 <= 3` |
| `>=` | 大於等於 | `3 >= 3` |

> `=` 用於指派值，`==` 用於判斷是否相等。

````python
print(3 == 3)
print(3 != 3)
print(3 < 3)
print(3 > 3)
print(3 <= 3)
print(3 >= 3)
````

## 3. 邏輯運算子

### 3.1 `and`：條件全部成立

可以想成**出國需要護照「而且」機票**，或遊戲需要**金鑰「而且」軟體**。只要有一個條件為 `False`，整體就是 `False`。

````python
print(True and True)
print(True and False)
print(False and True)
print(False and False)
````

### 3.2 `or`：至少一個條件成立

例如寫作業可以用**原子筆或鉛筆**；兩個條件都不成立時才是 `False`。

````python
print(True or True)
print(True or False)
print(False or True)
print(False or False)
````

### 3.3 `^`：布林值的 XOR（互斥或）

例如套餐飲料**可樂／玉米濃湯二選一**。對兩個布林值使用 `^` 時，**恰有一個為 `True`** 才會得到 `True`。

> Python 的 `^` 本質上是**位元 XOR** 運算子；這裡因為輸入為布林值，所以可用來展示邏輯上的互斥或。

````python
print(True ^ True)
print(True ^ False)
print(False ^ True)
print(False ^ False)
````

| A | B | `A and B` | `A or B` | `A ^ B` |
| :---: | :---: | :---: | :---: | :---: |
| `True` | `True` | `True` | `True` | `False` |
| `True` | `False` | `False` | `True` | `True` |
| `False` | `True` | `False` | `True` | `True` |
| `False` | `False` | `False` | `False` | `False` |

## 4. 運算優先順序

本課常見的順序是：**括號 → 比較運算 → `not` → `and` → `or`**。

例如 `True or False and True`，會先計算 `False and True`，再做 `or`。

> 不確定複合條件的解讀方式時，使用括號讓意圖更明確。

````python
print(True or False and True)  # True or (False and True) → True
````

### 多個條件組合

以下保留原本的逐步運算說明；練習時可先預測結果再執行。

````python
print(True or False and True) 
# -> True or False -> True

print(False and False or True)
# -> False or True -> True

print(True or False and True and False or True and False or True and True)
# -> True or (False and True and False) or (True and False) or (True and True)
# -> True or False or False or True
# -> True

print(1 == '1' or 1 >= 1 and False or True)
# -> False or True and False or True
# -> False or False or True
# -> True
````

## 5. 字串與複合指派

### 5.1 字串串接與換行

字串的 `+` 可串接內容，`"\n"` 代表換行字元。

````python
s1 = 'hello'
break_line = '\n'
s2 = 'hi'

print(s1 + break_line + s2)
````

### 5.2 複合指派運算子

`a += 2` 與 `a = a + 2` 效果相同（對此處的整數變數而言）。常見形式還有 `-=`、`*=`、`/=`、`//=`、`%=`。

注意：`+=` 屬於**指派**，而 `>=`、`<=`、`!=` 是**比較**。

````python
a = 10
a = a + 2
print(a)

a = 10
a += 2
print(a)
````

## 7. 綜合練習（5 題）

### 第 1 題｜長方形計算機

依序輸入長與寬，計算**面積**與**周長**。

- 面積：`length * width`
- 周長：`2 * (length + width)`
- 測試輸入：`3`、`4` → 面積 `12`、周長 `14`。

````python
length = int(input())
width = int(input())

area = length * width
perimeter = 2 * (length + width)

print('面積 =', area)
print('周長 =', perimeter)
````

### 第 2 題｜糖果包裝工廠

輸入糖果總數 `n`、每包糖果數 `k`，計算**完整包數**與**剩餘糖果**。

- `//`：整數除法（商）
- `%`：餘數
- 測試輸入：`8`、`5` → 完整包數 `1`、剩餘糖果 `3`。

> `k` 必須不等於 `0`，否則會產生除以零的錯誤。

````python
n = int(input())
k = int(input())

package = n // k 
remainder = n % k

print('完整包數 =', package)
print('剩餘糖果 =', remainder)
````

### 第 3 題｜學生獎勵資格判斷

依序輸入分數 `score` 與出席率 `attendance`：

- **活動資格**：分數至少 70 **且** 出席率至少 80
- **特別獎勵**：分數至少 95 **或** 出席率等於 100

執行後會分別顯示兩項判斷的布林結果。

````python
score = int(input())
attendance = int(input())

activity_score = score >= 70 and attendance >= 80
reward = score >= 95 or attendance == 100

print('活動資格 =', activity_score)
print('特別獎勵 =', reward)
````

### 第 5 題｜文字與清單魔法

練習字串的 `+`／`*` 與清單的 `+`／`*`。

- 字串 `+`：串接文字；字串 `*`：重複文字
- 清單 `+`：串接清單；清單 `*`：重複清單元素

````python
name = "Amy"
team1 = ["Amy", "Ben"]
team2 = ["Cindy", "David"]

print("Hello " + name)
print(name * 3)
print(team1 + team2)
print(team1 * 2)
````
