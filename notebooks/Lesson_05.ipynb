```python
## 比較運算子

print(3 == 3)
print(3 != 3)
print(3 < 3)
print(3 > 3)
print(3 <= 3)
print(3 >= 3)
```

```python
## and -> 
# 金鑰 還有 軟體 => Minecraft
# 出國 -> 護照 、 機票

print(True and True)
print(True and False)
print(False and True)
print(False and False)
```

```python
## or
# 寫作業 -> 原子筆 / 鉛筆
# 吃飯 -> 筷子 / 叉子

print(True or True)
print(True or False)
print(False or True)
print(False or False)
```

```python
# xor 
# 麥當勞的套餐 -> 漢堡 + 薯條 + 
# 可樂 / 玉米濃湯 -> 二擇一

print(True ^ True)
print(True ^ False)
print(False ^ True)
print(False ^ False)
```

```python
# 順序
# 有括號先算 → 比較運算 → and → or
# 3 > 2 -> 比較運算

print(True or False and True) # -> True or False -> True
```

```python
print(True or False and True) 
# -> True or False -> True

print(False and False or True)
# -> False or True -> True

print(True or False and True and False or True and False or True and True)
# -> True or (False and True and False) or (True and False) or (True and True)
# -> True or False or False or True
# -> True

print(1 == '1' or 1 >= 1 and False or True)
# -> False or True and False or True
# -> False or False or True
# -> True
```

```python
# 第 1 題｜長方形計算機

# length, width = input().split()
# length, width = int(length), int(width)

length = int(input())
width = int(input())

area = length * width
perimeter = 2 * (length + width)

print('面積 =', area)
print('周長 =', perimeter)
```

```python
# 第 2 題｜糖果包裝工廠

n = int(input())
k = int(input())

package = n // k 
remainder = n % k

print('完整包數 =', package)
print('剩餘糖果 =', remainder)
```

```python
# 第 3 題｜學生獎勵資格判斷

score = int(input())
attendance = int(input())

activity_score = score >= 70 and attendance >= 80
reward = score >= 95 or attendance == 100

print('活動資格 =', activity_score)
print('特別獎勵 =', reward)
```

```python
# 第 4 題｜遊戲分數更新

score = 100

score += 30 # 完成任務，增加 30 分
score -= 20 # 被怪物攻擊，扣除 20 分
score *= 2 # 獲得雙倍分數獎勵

print(score)
```
