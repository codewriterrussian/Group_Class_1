# Lesson 04｜Python 運算與二進位

本課介紹 Python 的基本運算子，以及十進位（base 10）與二進位（base 2）的相互轉換。

**學習目標**

- 認識四則運算：`+`、`-`、`*`、`/`。
- 理解次方 `**`、整除 `//` 與取餘數 `%`。
- 練習十進位與二進位的互換。

## 1. 基本四則運算

Python 可以直接使用運算子進行計算，例如計算金額。

| 運算 | Python 運算子 | 範例 | 結果 |
|:---|:---:|:---|---:|
| 加法 | `+` | `2 + 3` | `5` |
| 減法 | `-` | `5 - 2` | `3` |
| 乘法 | `*` | `2 * 3` | `6` |
| 除法 | `/` | `2 / 3` | 約 `0.6667` |

**注意：** Python 用 `*` 表示乘法，而 `/` 的結果可以是小數。

```python
print(2 + 3)
print(5 - 2)
print(2 * 3)
print(2 / 3)
```

## 2. 其他常用運算子

| 運算 | Python 運算子 | 範例 | 結果 |
|:---|:---:|:---|---:|
| 次方 | `**` | `2 ** 2` | `4` |
| 整除（向下取整） | `//` | `20 // 3` | `6` |
| 取餘數 | `%` | `20 % 3` | `2` |

### 次方

`2 ** 2` 是 \(2^2\)，`3 ** 3` 是 \(3^3\)。下列程式用 `==` 比較兩邊是否相等，回傳布林值 `True` 或 `False`。

**提醒：** Python 中 `^` 不是次方運算子，次方要用 `**`。

```python
print(2 ** 2 == 2 * 2)       # 4 == 4 → True
print(3 ** 3 == 3 * 3 * 3)   # 27 == 27 → True
```

### 整除與餘數

例如：\(20 = 3 \times 6 + 2\)，所以 **20 除以 3，商為 6、餘數為 2**。

- `//`：取得整除後的整數商（對正整數而言）。
- `%`：取得餘數。

```python
print(20 // 3)  # 6：商
print(20 % 3)   # 2：餘數
```

## 3. 十進位 → 二進位（10 → 2）

二進位只使用 `0` 和 `1`。從右往左，每一位分別代表 \(2^0, 2^1, 2^2, 2^3, \ldots\)，也就是 **1、2、4、8、16、32、64、128、256、512……**。

轉換方式：找出能組成原數字的 2 的次方；有用到的位數記為 `1`，沒有用到的記為 `0`。

| 十進位 | 拆成 2 的次方 | 二進位 |
|---:|:---|:---|
| `234` | `128 + 64 + 32 + 8 + 2` | `11101010` |
| `512` | `512` | `1000000000` |
| `24` | `16 + 8` | `11000` |
| `9` | `8 + 1` | `1001` |

**例：234** 需要 128、64、32、8、2；依序在對應的位數放 `1`，其餘放 `0`，得到 `11101010`。

## 4. 二進位 → 十進位（2 → 10）

讀取二進位中所有為 `1` 的位置，把它們對應的 2 的次方加總。

| 二進位 | 加總 | 十進位 |
|:---|:---|---:|
| `10101` | `16 + 4 + 1` | `21` |
| `11` | `2 + 1` | `3` |
| `11000` | `16 + 8` | `24` |
| `1100101` | `64 + 32 + 4 + 1` | `101` |

**例：10101₂**：從右往左數，第 0、2、4 位為 `1`，所以 \(2^0 + 2^2 + 2^4 = 1 + 4 + 16 = 21\)。

## 5. 延伸：用 Python 檢查二進位轉換

這一節是補充工具：

- `bin(數字)`：將十進位整數轉成二進位字串（前面的 `0b` 表示二進位）。
- `int("二進位字串", 2)`：將二進位字串轉成十進位整數。

```python
# 十進位 → 二進位
for number in (234, 512, 24, 9):
    print(f"{number} → {bin(number)}")

print()  # 空一行

# 二進位 → 十進位
for binary in ("10101", "11", "11000", "1100101"):
    print(f"{binary} → {int(binary, 2)}")
```

## 本課重點整理

- **四則運算：** `+`、`-`、`*`、`/`
- **次方：** `**`；**整除：** `//`；**餘數：** `%`
- **十進位轉二進位：** 將數字拆成不同的 2 的次方。
- **二進位轉十進位：** 將為 `1` 的位數所代表的數值相加。
