## 2｜傳統 `%` 格式化

- `%d`：整數
- `%f`：浮點數；`%.2f` 顯示小數點後兩位
- `%s`：字串

可使用 `% (值1, 值2, …)` 一次填入多個值。

````python
# 浮點數、整數、字串

temp = 24
hum = 60

print('今天天氣 24 度， 濕度 60')

print("%d" % 20) # decimal
print("%.2f" % 20)
print("%s" % 'Hi')

print("%d %s %.2f" % (20, "Hi", 3.5))
````

````python
temp = 32.2
hum = 32

print('今天天氣 %.2f 度， 濕度 %.1f' % (temp, hum))
````

### 欄位寬度與小數位數

例如 `%10.2f` 表示**最小欄位寬度 10**、小數點後 2 位；不足寬度時預設靠右對齊。

````python
num = 5

print(num)
print("%.2f" % num)
print("%3.2f" % num)
print("%5.2f" % num)
print("%10.2f" % num)
````

## 3｜f-string 格式化

在字串前加 `f`，以 `{變數}` 插入值；例如 `{score:.2f}`、`{price:,.2f}`、`{hour:02d}`。

````python
# f-string

temp = 32
hum = 32

print(f'今天天氣 {temp:03d} 度， 濕度 {hum}')
# print('今天天氣 %.2f 度， 濕度 %.1f' % (temp, hum))
````

### 練習 1｜姓名、年齡與分數

練習將分數顯示到小數點後兩位。

**題目與預期輸出**

```text
name = "Alice"
age = 18
score = 95.678

預期輸出：

```

````python
name = "Alice"
age = 18
score = 95.678

print(f"Name: {name}\nAge: {age}\nScore: {score:.2f}")
````

### 練習 2｜商品價格與總額

練習千分位與金額的小數位數。

**題目與預期輸出**

```text
price = 1234567.891
quantity = 3

預期輸出：
Price: $1,234,567.89
Quantity: 3

```

````python
price = 1234567.891
quantity = 3
total = price * quantity


print(f"Price: ${price:,.2f}\nQuantity: {quantity}\nTotal: ${total:,.2f}")
````

### 練習 3｜時間格式

**題目與預期輸出**


````python
hour = 8
minute = 5
second = 9

print(f"Current time: {hour:02d}:{minute:02d}:{second:02d}")
````

### 作業 1｜Shopping Receipt

**已知變數**

```python
product = "Keyboard"
```

**預期輸出**


````python
# TODO：請在此撰寫程式

product = "Keyboard"
price = 1580.5
quantity = 3
discount = 0.15

print('===== Shopping Receipt =====')
print(f'Product: {product}\n' + 
      f'Unit Price: ${price:,.2f}\n' +
      f'Quantity: {quantity}\n' +
      f'Subtotal: ${price * quantity:,.2f}\n' +
      f'Discount: ${discount * 100}%\n' +
      f'Final Total: ${price * quantity * (1-discount):,.2f}'
      )
print('============================')
````

### 作業 2｜Student Report

**已知變數**

```python

```

**預期輸出**


````python
# TODO：請在此撰寫程式
name = "Michael"
chinese = 87
english = 92
math = 78
avg = (chinese+english+math)/3

if avg >= 60:
    pass_text = 'PASS'
else:
    pass_text = 'FAIL'


print('===== Student Report =====')

print(f'Name: {name}\n\n' + 
      f'Subject\tScore\n' +
      '------------------\n' +
      f'Chinese\t{chinese}\n' +
      f'English\t{english}\n' +
      f'Math\t{math}\n' +
      '------------------\n' +
      f'Total:\t{chinese+english+math}\nAverage:\t{avg:.2f}\n'+
      f'Result:\t{pass_text}'
      )
print('==========================')
````

## 7｜重點整理

| 語法 | 用途 | 範例 |
|---|---|---|
| `\n` | 換行 | `"A\nB"` |
| `\t` | Tab | `"A\tB"` |
| `%d` | 整數 | `"%d" % 5` |
| `%.2f` | 小數兩位 | `"%.2f" % 3.5` |
| `{x:.2f}` | f-string 小數兩位 | `f"{x:.2f}"` |
| `{x:,.2f}` | 千分位 | `f"{x:,.2f}"` |
| `{x:02d}` | 整數補零 | `f"{x:02d}"` |
| `sep=""` | 移除參數間空格 | `print("A", "B", sep="")` |

````python
x = 2
print(f'{{x}} 的結果是 {x}')

print('\\n 會 \n 跳行')

print('%%s 是 %s' % 'hi')

print('%s' % 'hi')

print('%')
````
